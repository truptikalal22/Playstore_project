```python
# diffrent size to one size(kb) convert
def convert_size(x):
    x = str(x).strip().upper()
    x = x.replace(',','')

    if x.endswith('K'):
        return float(x[:-1])

    elif x.endswith('M'):
        return float(x[:-1]) * 1024

    elif x.endswith('G'):
        return float(x[:-1]) * 1024 * 1024

    else:
        return np.nan

df['Size'] = df['Size'].apply(convert_size)
```

```python
# pontential outlier nikalne k liye

Q1 =  df['Installs'].quantile(0.25)
Q3 =  df['Installs'].quantile(0.75)
```

```python
IQR = Q3 - Q1
print("IQR:",IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
outliers =df[ (df['Installs'] < lower_limit) | (df['Installs'] > upper_limit)]
```

```python
outliers.head()
```

```python
len(outliers)
```

```python
# pontential outlier all colmns
cols = ['Installs', 'Rating Count', 'Price', 'Size']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f'{col}: {len(outliers)} outliers')
```
